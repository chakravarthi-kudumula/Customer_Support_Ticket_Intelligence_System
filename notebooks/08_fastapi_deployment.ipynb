# Phase 8: FastAPI Deployment

This notebook shows the local API contract for classification, search, RAG-style retrieval, summarization, and combined analysis.


## Run Server

```bash
uvicorn src.api.main:app --reload --host 127.0.0.1 --port 8000
```


````python
from fastapi.testclient import TestClient
from src.api.main import app

client = TestClient(app)
client.get("/health").json()
````

````python
client.post("/classify", json={"text": "My account was charged twice"}).json()
````

````python
client.post("/rag", json={
    "query": "My account was charged twice",
    "top_k": 3,
    "filters": {"product": "Checking or savings account"},
}).json()
````

## Notes

- `/summarize` lazy-loads the seq2seq model, so the first call is slower.
- `/search` and `/rag` use the full 90k FAISS index.
- `/analyze` combines classification and retrieval, and can optionally include summarization.
